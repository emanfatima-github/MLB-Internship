```python
from ultralytics import YOLO
import os
import cv2
import matplotlib.pyplot as plt
import pandas as pd

# 1. Load V1 and V2

V1_MODEL = r"C:\Users\Localws\Bottle-Detection-1\best.pt"
V2_MODEL = r"C:\Users\Localws\Bottle-Detection-1\Day37_Training\V2\weights\best.pt"
DATA_YAML = r"C:\Users\Localws\Bottle-Detection-1\data.yaml"

model_v1 = YOLO(V1_MODEL)
model_v2 = YOLO(V2_MODEL)


# 2. Evaluate V1 and V2

v1 = model_v1.val(data=DATA_YAML, split="val")
v2 = model_v2.val(data=DATA_YAML, split="val")

comparison = pd.DataFrame({
    "Metric": ["Precision", "Recall", "mAP@50", "mAP@50-95"],
    "V1": [
        v1.box.mp,
        v1.box.mr,
        v1.box.map50,
        v1.box.map
    ],
    "V2": [
        v2.box.mp,
        v2.box.mr,
        v2.box.map50,
        v2.box.map
    ]
})

print("V1 vs V2 Comparison")
display(comparison.round(4))

# 3. Select 5 test images

test_images_path = r"C:\Users\Localws\Bottle-Detection-1\test\images"

test_images = [
    os.path.join(test_images_path, f)
    for f in os.listdir(test_images_path)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

test_images = test_images[:5]

# 4. Compare V1 and V2

for i, image_path in enumerate(test_images, 1):

    v1_result = model_v1.predict(
        source=image_path,
        conf=0.25,
        verbose=False
    )[0]

    v2_result = model_v2.predict(
        source=image_path,
        conf=0.25,
        verbose=False
    )[0]

    v1_image = cv2.cvtColor(v1_result.plot(), cv2.COLOR_BGR2RGB)
    v2_image = cv2.cvtColor(v2_result.plot(), cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(v1_image)
    plt.title(f"Example {i} - V1")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(v2_image)
    plt.title(f"Example {i} - V2")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
